```python
import numpy as np

x = np.array([1, 2, 3, 4, 5, 6])

condlist = [x < 5, x > 5]
choicelist = ['Less than 5', 'Greater than 5']

result = np.select(condlist, choicelist, default='Equal to 5')

print(result)
```
